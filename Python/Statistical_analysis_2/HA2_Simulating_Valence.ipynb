### Exercise 1. (*1pt*)
Load the data set as provided in the file `ANEW_Valence_per_word.tsv` into the variable `per_word`.

#### Solution

```python
per_word = pd.read_csv("ANEW_Valence_per_word.tsv", sep='\t')
per_word
```

### Exercise 2. (*2pt*)
The score per document can be calculated as the average score of all observed words per document. Plot the distribution of the observed sentiment distribution for all documents (recall that we can observe values ranging from 1 to 9).

#### Solution

```python
a = per_word.groupby('doc_id')['ANEW_Valence'].transform (np.mean)
fig = plt.figure(figsize=(60, 60))
ax = fig.add_subplot()
a.hist(ax=ax, bins=np.arange(1, 9, 0.1), density=True, facecolor=(*tab10[0], 0.6), edgecolor="k")
ax.set_ylabel("Frequency")
ax.set_xlim(1, 9)
ax.set_xlabel("Sentiment score");
```

### Exercise 3. (*2pt*)
Plot the distribution of the number of words that are contained in the lexicon per document.

#### Solution

```python
b = per_word.groupby('doc_id')['Word'].nunique()
fig = plt.figure(figsize=(60, 60))
ax = fig.add_subplot()
b.hist(ax=ax, bins=np.arange(0, 267, 1), density=True, facecolor=(*tab10[0], 0.6), edgecolor="k")
ax.set_ylabel("Number of words in the lexicon per document")
ax.set_xlim(0, 267)
ax.set_xlabel("Sentiment score");
```

### Exercise 4. (*2pt*)
Plot the distribution of the observed sentiment distribution for all words (recall that we can observe values ranging from 1 to 9).

#### Solution

```python
c = per_word["ANEW_Valence"]
fig = plt.figure(figsize=(60, 60))
ax = fig.add_subplot()
c.hist(ax=ax, bins=np.arange(0, 9, 0.1), density=True, facecolor=(*tab10[0], 0.6), edgecolor="k")
ax.set_ylabel("Frequency")
ax.set_xlim(0, 9)
ax.set_xlabel("Sentiment score");
```
